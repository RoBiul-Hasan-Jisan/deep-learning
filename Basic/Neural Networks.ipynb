# Introduction to Neural Networks

## 1. What is a Neural Network?
A **Neural Network (NN)** is a computational model inspired by the human brain. It is made up of interconnected neurons (nodes) connected by weights, allowing it to learn complex patterns from data. 

Think of it as a system that: **Takes Input $\rightarrow$ Processes It $\rightarrow$ Produces Output**.

> **Key Idea:** Neural networks are exceptionally good at pattern recognition, making them the standard for tasks involving images, sound, and complex text.






### 1.1 Basic Terminology
* **Neuron (Node):** A single unit that receives input, processes it, and produces an output.
* **Layer:** A collection of neurons operating at the same stage.
  * **Input Layer:** Takes the raw data ( pixels of an image or tabular features).
  * **Hidden Layer(s):** Processes the features. Networks with many hidden layers are called "Deep" Neural Networks.
  * **Output Layer:** Produces the final prediction ( a class label or a continuous number).
* **Weight ($w$):** Determines the importance of a specific input connection.
* **Bias ($b$):** A constant added to the neuron to shift the activation function, helping it better fit the data.
* **Activation Function:** A mathematical equation that adds non-linearity, allowing the network to learn highly complex, non-linear patterns.

---

### 1.2 How a Single Neuron Works
A single neuron computes a weighted sum of its inputs, adds a bias, and passes the result through an activation function.

**The Linear Combination:**
$$z = w_1x_1 + w_2x_2 + \dots + w_nx_n + b$$

**The Activation:**
$$y = f(z)$$

**Where:**
* $x_1, x_2, \dots, x_n \rightarrow$ Inputs
* $w_1, w_2, \dots, w_n \rightarrow$ Weights
* $b \rightarrow$ Bias
* $f(z) \rightarrow$ Activation function
* $y \rightarrow$ Final Output



**Common Activation Functions:**
* **Sigmoid:** $f(z) = \frac{1}{1 + e^{-z}}$
  * *Outputs between 0 and 1. Good for binary probabilities.*
* **ReLU (Rectified Linear Unit):** $f(z) = \max(0, z)$
  * *Outputs 0 if negative, otherwise outputs the raw value. The default standard for hidden layers.*
* **Tanh (Hyperbolic Tangent):** $f(z) = \tanh(z)$
  * *Outputs between -1 and 1. Zero-centered.*
* **Softmax:** * *Converts a vector of raw outputs into a probability distribution. Used in the output layer for multi-class classification.*

---

## 2. Architecture of Neural Networks
A neural network is organized sequentially into layers:

1. **Input Layer:** The number of neurons equals the number of features in your dataset.
2. **Hidden Layers:** Can be shallow (1-2 layers) or deep (dozens of layers). Each layer contains neurons with activation functions.
3. **Output Layer:** The number of neurons depends entirely on the task:
   * **Regression:** 1 neuron (Linear activation).
   * **Binary Classification:** 1 neuron (Sigmoid activation).
   * **Multi-class Classification:** $N$ neurons (Softmax activation, where $N$ is the number of classes).

*A simplified visualization:*
```text
Input Layer      Hidden Layer        Output Layer
  x1  ----\
  x2  ----- [Neuron] ---\
  x3  ----/              [Neuron] --> y

```

## 3. How Neural Networks Learn
Learning is simply the process of adjusting weights and biases to minimize prediction errors.

### 3.1 Forward Propagation
Information flows forward through the network to make a guess:
1. Pass inputs in.
2. Multiply by weights, add bias, apply activation.
3. Repeat for all layers to get the final output prediction.

### 3.2 Loss Function
Measures exactly how wrong the network's prediction is compared to the true label.
* **Mean Squared Error (MSE):** Used for Regression.
* **Binary Cross-Entropy:** Used for Binary Classification.
* **Categorical Cross-Entropy:** Used for Multi-class Classification.

### 3.3 Backpropagation & Gradient Descent
To improve predictions, the network must update its weights backwards based on the error.



1. Calculate the gradient (slope) of the loss function with respect to the weights using calculus (the Chain Rule).
2. Update the weights to step "downhill" toward the minimum error:
   $$w_{\text{new}} = w_{\text{old}} - (\text{learning\_rate} \times \text{gradient})$$
3. Repeat this process for many iterations (epochs) until the loss is minimized.



### 3.4 Gradient Descent Variants
* **SGD (Stochastic Gradient Descent):** Updates weights using one single example at a time.
* **Mini-batch Gradient Descent:** Updates weights using a small batch of data (the industry standard).
* **Adam:** An advanced optimizer with an adaptive learning rate; highly recommended for most modern networks.

---

## 4. Types of Neural Networks
* **Feedforward Neural Network (FNN):** Simple, traditional architecture. No cycles; information only flows forward.
* **Convolutional Neural Network (CNN):** Specialized for spatial data like **images** and video.
* **Recurrent Neural Network (RNN):** Specialized for sequential data like **time series** or text.
* **Transformers:** The modern state-of-the-art architecture for NLP (like ChatGPT) and advanced vision tasks.

## 5. Practical Example (Binary Classification)
**The Problem:** Predict if a person wants to eat ice cream based on their age and the outside temperature.
* **Inputs:** Age ($x_1$), Temperature ($x_2$)
* **Target Output:** `0` (No) or `1` (Yes)

**The Process:**
1. **Architecture:** Input Layer (2 neurons) $\rightarrow$ Hidden Layer (3 neurons, ReLU) $\rightarrow$ Output Layer (1 neuron, Sigmoid).
2. **Forward Pass:** The data passes through the network to predict a probability ( `0.85`).
3. **Compute Loss:** Compare the prediction (`0.85`) to the actual truth (`1.0`).
4. **Backpropagation:** Adjust the weights based on the error.
5. **Result:** After training on many examples, the network successfully learns complex, non-linear patterns, such as: *"Young people are highly likely to eat ice cream, especially when the temperature is hot."*
