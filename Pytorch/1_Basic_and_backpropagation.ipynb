## Tensor basics

### Creating Tensors

```python
import torch

# From list
t1 = torch.tensor([1, 2, 3])

# Zeros / Ones
t2 = torch.zeros(3, 3)
t3 = torch.ones(2, 2)

# Random
t4 = torch.rand(2, 3)

# Range
t5 = torch.arange(0, 10)

print(t1, t2, t3, t4, t5)
```

### Tensor Shape

```python
t = torch.rand(3, 4)
print(t.shape)   # (3, 4)
print(t.size())  # same as shape
```

### Tensor Data Types

```python
t = torch.tensor([1, 2, 3], dtype=torch.float32)

print(t.dtype)
```

### Math Scalar Operations

```python
t = torch.tensor([1, 2, 3])

print(t + 5)
print(t * 2)
print(t / 2)
```

### Element-wise Operations

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

print(a + b)
print(a * b)
```

### Reduction Operations

```python
t = torch.tensor([1.0, 2.0, 3.0])

print(torch.sum(t))
print(torch.mean(t))
print(torch.max(t))
```

### Matrix Operations

```python
A = torch.rand(2, 3)
B = torch.rand(3, 2)

# Matrix multiplication
C = torch.matmul(A, B)

print(C)
```

### Comparison Operations

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([2, 2, 2])

print(a > b)
print(a == b)
```

### Special Functions

```python
t = torch.tensor([1.0, 4.0, 9.0])

print(torch.sqrt(t))
print(torch.exp(t))
print(torch.log(t))
```

### In-place Operations

```python
t = torch.tensor([1, 2, 3])

t.add_(5)   # inplace
print(t)
```

### Copying a Tensor

```python
a = torch.tensor([1, 2, 3])

b = a.clone()   # safe copy
b[0] = 100

print(a)  # unchanged
print(b)
```

## Backpropagation

**Formula:**

* Forward → compute output
* Loss → error
* Backward → gradients
* Update weights

### Weight Initialization

```python
w = torch.randn(2, 2, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
```

### Forward Propagation

```python
x = torch.tensor([[1.0, 2.0]])

y = x @ w + b   # linear layer
```

### Weight Update (Gradient Descent)

```python
loss = y.sum()

loss.backward()

# Update
with torch.no_grad():
    w -= 0.01 * w.grad
    b -= 0.01 * b.grad

    w.grad.zero_()
    b.grad.zero_()
```

```python
import torch

# Data
x = torch.tensor([[1.0, 2.0]])
y_true = torch.tensor([[1.0]])

# Weights
w = torch.randn(2, 1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)

# Forward
y_pred = x @ w + b

# Loss
loss = (y_pred - y_true)**2

# Backward
loss.backward()

# Update
with torch.no_grad():
    w -= 0.01 * w.grad
    b -= 0.01 * b.grad

    w.grad.zero_()
    b.grad.zero_()

print("Updated weights:", w)
```
