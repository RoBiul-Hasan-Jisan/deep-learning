```python
import numpy as np
```

```python
# Input = 2, Hidden = 3, Output = 1
np.random.seed(0)

W1 = np.random.randn(2, 3)
b1 = np.zeros((1, 3))

W2 = np.random.randn(3, 1)
b2 = np.zeros((1, 1))
```

```python
X = np.array([[1, 2]])   # 1 sample, 2 features
y = np.array([[1]])      # target
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
# Hidden layer
Z1 = np.dot(X, W1) + b1
A1 = sigmoid(Z1)

# Output layer
Z2 = np.dot(A1, W2) + b2
A2 = sigmoid(Z2)

print("Prediction:", A2)
```

```python
loss = np.mean((A2 - y) ** 2)
print("Loss:", loss)
```

```python
# Output layer gradient
dA2 = A2 - y
dZ2 = dA2 * sigmoid_derivative(A2)

dW2 = np.dot(A1.T, dZ2)
db2 = np.sum(dZ2, axis=0, keepdims=True)

# Hidden layer gradient
dA1 = np.dot(dZ2, W2.T)
dZ1 = dA1 * sigmoid_derivative(A1)

dW1 = np.dot(X.T, dZ1)
db1 = np.sum(dZ1, axis=0, keepdims=True)
```

```python
lr = 0.1

W1 -= lr * dW1
b1 -= lr * db1

W2 -= lr * dW2
b2 -= lr * db2
```

```python
for i in range(1000):

    # Forward
    Z1 = np.dot(X, W1) + b1
    A1 = sigmoid(Z1)

    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)

    # Loss
    loss = np.mean((A2 - y) ** 2)

    # Backward
    dA2 = A2 - y
    dZ2 = dA2 * sigmoid_derivative(A2)

    dW2 = np.dot(A1.T, dZ2)
    db2 = np.sum(dZ2, axis=0, keepdims=True)

    dA1 = np.dot(dZ2, W2.T)
    dZ1 = dA1 * sigmoid_derivative(A1)

    dW1 = np.dot(X.T, dZ1)
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    # Update
    W1 -= lr * dW1
    b1 -= lr * db1
    W2 -= lr * dW2
    b2 -= lr * db2

    if i % 100 == 0:
        print(f"Epoch {i}, Loss: {loss}")
```
