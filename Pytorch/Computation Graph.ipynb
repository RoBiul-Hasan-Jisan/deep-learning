#  Autograd (Automatic Differentiation)

## 1. Introduction
In deep learning, we need to compute gradients automatically to update weights and train models. That is exactly where **Autograd** (Automatic Differentiation) comes in.

Instead of mathematically calculating derivatives by hand for every single architecture, the system:
1. Builds a computation graph in the background.
2. Applies the calculus chain rule automatically.
3. Computes the exact gradients highly efficiently.

---

## 2. Why Autograd?
Imagine a modern neural network like GPT with billions of parameters. 

Manually computing the gradient of the Loss with respect to a single weight:
$$\frac{\partial L}{\partial w}$$

Is practically impossible because it is:
*  **Highly Error-prone:** A single missed negative sign ruins the entire network.
*  **Impossible at Scale:** You cannot hard-code calculus for millions of changing parameters.

**Autograd solves this entirely by:**
* Tracking every mathematical operation you do.
* Automatically computing the gradients backwards.
* Enabling seamless backpropagation.

---

## 3. The Computation Graph
A computation graph is a directed mathematical graph where:
* **Nodes:** Represent variables or mathematical operations (like addition or multiplication).
* **Edges:** Represent the flow of data/computation.
