# Decision Boundaries, Deep Learning Types & Visualization Tools

## What is a Decision Boundary?

In classification problems, a **Decision Boundary** is the line (in 2D), plane (in 3D), or hyperplane (in higher dimensions) that separates different classes of data.

* **The Goal:** The neural network tries to learn the optimal boundary that cleanly separates "Class A" from "Class B".
* **Linear Boundary:** A straight line. Produced by simple models like Logistic Regression or a Single Perceptron.
* **Non-Linear Boundary:** A curved or complex shape. Produced by Deep Neural Networks with non-linear activation functions (ReLU, Tanh).

## Types of Deep Learning Architectures

Deep Learning is not just one thing; it's a family of architectures designed for specific types of data.

### A. Artificial Neural Networks (ANN) / Multi-Layer Perceptrons (MLP)
* **Structure:** Dense layers where every neuron is connected to every neuron in the next layer.
* **Use Case:** Tabular data (Excel sheets), simple classification, regression.
* **Strength:** Good baseline for general problems.

### B. Convolutional Neural Networks (CNN)
* **Structure:** Uses "filters" (kernels) to scan inputs. Preserves spatial relationships.
* **Use Case:** **Image processing**, video analysis, medical imaging.
* **Strength:** Automatically detects edges, textures, and shapes.




### C. Recurrent Neural Networks (RNN) & LSTMs
* **Structure:** Has a "memory" loop where output from the previous step is fed as input to the current step.
* **Use Case:** **Sequential data** (Time series, text/NLP, audio, stock prices).
* **Strength:** Understands context and temporal order.

### D. Generative Adversarial Networks (GANs)
* **Structure:** Two networks fighting each other—a **Generator** (creates fake data) and a **Discriminator** (detects fake data).
* **Use Case:** Generating realistic images, art, deepfakes, drug discovery.

##  Decision Boundary & Bias Visualization

Why do we need a **Bias** term ($b$) in the equation $z = w \cdot x + b$?

### The Geometry of Bias
* **Without Bias ($z = w \cdot x$):** The decision boundary **must pass through the origin (0,0)**.
    * *Problem:* If your data is not centered at (0,0), the model cannot separate it.
* **With Bias ($z = w \cdot x + b$):** The decision boundary can shift **up, down, left, or right**. It is no longer tethered to the origin.



### Visualizing the Shift
1.  **$b = 0$:** The line crosses (0,0).
2.  **$b > 0$:** The line shifts towards the positive side (or up/left depending on axis).
3.  **$b < 0$:** The line shifts towards the negative side.

> **Analogy:** Bias is the "intercept" in the equation of a line ($y = mx + c$). Without $c$, you can only rotate the line, not move it.

##  Perceptron Learning Rule (Part 1: The Mechanism)

The "Learning Rule" is the algorithm that moves the decision boundary.

### The Logic Flow
1.  **Initialize:** Start with random weights (e.g., $w = [0.1, -0.2]$).
2.  **Predict:** Feed an input $x$. Get output $\hat{y}$.
3.  **Check:** Is $\hat{y}$ equal to the actual target $y$?
    * **Yes:** Do nothing. (The boundary is fine for this point).
    * **No:** Update the weights.

### The "Nudge"
When the prediction is wrong, we nudge the weight vector.
* **If we predicted 0 but answer is 1:** We need more "positive" signal.
    * $\text{New Weight} = \text{Old Weight} + (\text{Input} \times \text{Learning Rate})$
* **If we predicted 1 but answer is 0:** We need less "positive" signal.
    * $\text{New Weight} = \text{Old Weight} - (\text{Input} \times \text{Learning Rate})$

This simple addition/subtraction rotates the decision boundary slightly until the error is corrected.

## TensorFlow Playground

**TensorFlow Playground** is an interactive web-based tool created by Google to visualize how neural networks learn without writing code.

* **URL:** [playground.tensorflow.org](https://playground.tensorflow.org)



### How to use it (Mini-Lab):

**Experiment 1: The Power of Depth**
1.  **Select Data:** Choose the "Circle" dataset (blue dots inside, orange dots outside).
2.  **Setup:** Use 0 Hidden Layers.
3.  **Run:** Notice it fails. A single layer can only draw a straight line.
4.  **Modify:** Add **1 Hidden Layer** with 3 neurons. Use **ReLU** activation.
5.  **Run:** Watch the boundary bend into a polygon/circle to enclose the blue dots.

**Experiment 2: The Spiral (The Hardest Problem)**
1.  **Select Data:** Choose the "Spiral" dataset.
2.  **Challenge:** Try to classify this with 1 hidden layer. (It will likely fail).
3.  **Solution:** Add 2 or 3 hidden layers and change Activation to **Tanh** or **ReLU**.
4.  **Observe:** See how the network "folds" the space to separate the two spirals.

### Key Controls to Watch:
* **Epochs:** How many times the network has seen the data.
* **Learning Rate:** If too high, the loss jumps around. If too low, it learns too slowly.
* **Regularization:** Used to prevent overfitting (making the boundary too jagged).
