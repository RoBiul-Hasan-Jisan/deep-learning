# Artificial Neural Network 

## 1. What is an Artificial Neural Network (ANN)?
An Artificial Neural Network is a computational model inspired by the human brain, designed to recognize patterns and solve complex problems.

### Biological Inspiration
* **The Brain:** Inspired by the brain’s biological network of neurons.
* **The Mechanism:** Just like biological neurons pass electrical signals to one another, ANN neurons pass numerical signals.

### Structure of an ANN


* **Neurons (Nodes):** Small computational units that take inputs, apply weights and a bias, pass the result through an activation function, and produce an output.
* **Layers:**
    * **Input Layer:** Takes the raw features (data).
    * **Hidden Layers:** Process the data and extract complex patterns.
    * **Output Layer:** Produces the final predictions ( a regression value or classification probabilities).
* **Connections:** Each neuron connects to neurons in the next layer with **weights** that determine the mathematical importance of that specific connection.

---

## 2. How Does an ANN Learn?

###  Forward Propagation
Data flows strictly from input $\rightarrow$ hidden layers $\rightarrow$ output.
Each neuron computes a linear sum and then applies an activation function:
$$z = \sum(w_i \cdot x_i) + b$$

###  Loss Function
Measures exactly how wrong the network's predictions are compared to the true labels.
* **Examples:** MSE (for regression), Cross-Entropy (for classification).

###  Backpropagation
Calculates the gradient (sensitivity) of the loss with respect to each weight using the **Chain Rule** from calculus.

###  Gradient Descent / Optimizers
Updates the weights in the opposite direction of the gradient to minimize the loss.
* **Note:** Variants like **Adam** adapt the learning rates automatically for faster convergence.

## 3. Activation Functions

| Function | Formula / Output Range | When to Use | Notes |
| :--- | :--- | :--- | :--- |
| **Sigmoid** | $f(x) = \frac{1}{1+e^{-x}} \rightarrow (0, 1)$ | Binary classification output | Prone to vanishing gradients |
| **Tanh** | $f(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} \rightarrow (-1, 1)$ | Centered data; hidden layers | Still can vanish for large inputs |
| **ReLU** | $f(x) = \max(0, x)$ | Most hidden layers in deep nets | Avoids vanishing gradient; simple |
| **Leaky ReLU** | $f(x) = \begin{cases} x & \text{if } x > 0 \\ \alpha x & \text{else} \end{cases}$ | Hidden layers when ReLU “dies” | Keeps small gradient for negatives |

###  The Vanishing Gradient Problem
* With **Sigmoid/Tanh**, gradients become microscopically small for large absolute inputs ($|x|$). This slows or completely stops learning in early layers.
* **ReLU** is often preferred because it keeps gradients alive (a constant slope of `1`) for all positive values.

## 4. Loss Functions

| Type | Common Loss | Used For |
| :--- | :--- | :--- |
| **Regression** | MSE (Mean Squared Error) | Predicting continuous values ( prices, temperatures) |
| **Classification**| Cross-Entropy | Predicting class probabilities ( Cat vs. Dog) |

**Criteria for a Good Loss Function:**
1. Differentiable (required for backpropagation).
2. Represents your goal metric well.
3. Penalizes errors appropriately.

## 5. Hyperparameters & Initialization
Hyperparameters are the settings you configure *before* training begins.

* **Learning Rate ($\alpha$):** The step size taken in gradient descent.
* **Batch Size:** The number of samples processed before the model updates its weights.
* **Epochs:** The number of complete passes through the entire training dataset.
* **Model Capacity:** The number of hidden layers and neurons.

**Weight Initialization:**
* **Xavier (Glorot):** Provides balanced gradients; best suited for **Sigmoid/Tanh**.
* **He Initialization:** Accounts for sparsity; best suited for **ReLU**.

## 6. Overfitting vs. Underfitting



| Problem | Symptom | Fix |
| :--- | :--- | :--- |
| **Overfitting** | High train accuracy, low test accuracy | Regularization (L1/L2), Dropout, Early stopping, More data |
| **Underfitting**| Low train & test accuracy | Increase model size, Train longer, Better features |

### Core Solutions:
* **Regularization (L1/L2):** * **L1:** Encourages sparsity (pushes weights to exactly `0`).
    * **L2:** Prevents excessively large weights.
* **Dropout:** Randomly “turns off” a percentage of neurons during training to prevent the network from relying too heavily on specific nodes.
* **Early Stopping:** Halts training exactly when the validation loss stops improving to prevent the model from memorizing the noise.

## 7. Deep Learning Q&A (Interview Prep)

**Q1: What is the difference between ANN, CNN, and RNN?**
* **ANN (Artificial Neural Network):** General-purpose network where each neuron connects to all neurons in the next layer. Best for tabular/non-sequential data.
* **CNN (Convolutional Neural Network):** Specialized for spatial data (images, video). Uses convolution layers to detect local patterns (edges) and pooling layers for dimensionality reduction.
* **RNN (Recurrent Neural Network):** Designed for sequential data (text, time series). Has loops to retain information from previous steps. LSTMs/GRUs solve its long-term memory issues.



**Q2: Why does ReLU work better than Sigmoid?**
* Sigmoid saturates for large positive/negative inputs, leading to vanishing gradients. 
* **ReLU** ($f(x) = \max(0, x)$) does not saturate in the positive region, ensuring strong gradients and faster convergence. It is also computationally much cheaper.

**Q3: Explain Backpropagation in simple terms.**
* **Forward pass:** Calculate outputs and error (loss).
* **Backward pass:** Compute how much each weight contributed to the error (using the chain rule). Adjust each weight slightly in the opposite direction of the gradient. *Think of it as “blaming” each weight for its share of the error, then “correcting” it.*

**Q4: How would you prevent overfitting in a neural network?**
* Use **Regularization** (L1/L2 penalties).
* Apply **Dropout** layers.
* Use **Data Augmentation** (rotate, flip, scale images; add noise).
* Implement **Early Stopping**.
* Reduce model complexity (fewer layers/neurons).
* Use Cross-Validation.

**Q5: What happens if the learning rate is too high/low?**
* **Too high:** The optimizer overshoots the minimum, oscillating wildly or diverging entirely.
* **Too low:** The optimizer converges painfully slowly and can get permanently stuck in local minima.

**Q6: How do you choose the number of hidden layers and neurons?**
* There is no fixed mathematical rule; it depends on data complexity.
* Start simple (1–2 hidden layers for small problems) and scale up.
* Increase capacity until validation accuracy stops improving.
* Use Grid Search, Random Search, or AutoML to tune.

**Q7: How is MSE different from Cross-Entropy?**
* **MSE:** Measures the average squared difference between continuous predictions and actuals. Terrible for classification because it penalizes confidence linearly.
* **Cross-Entropy:** Measures the distance between probability distributions. Heavily penalizes models that are highly confident but entirely wrong.

**Q8: Explain Vanishing / Exploding gradients.**
* **Vanishing:** In deep networks, gradients shrink exponentially as they propagate backward, causing early layers to stop learning. Common with Sigmoid/Tanh.
* **Exploding:** Gradients grow uncontrollably, causing weights to blow up and destabilize training. Common in basic RNNs.
* **Solutions:** Use ReLU, Gradient Clipping (for exploding), proper initialization (Xavier/He), and architectures like LSTM.
