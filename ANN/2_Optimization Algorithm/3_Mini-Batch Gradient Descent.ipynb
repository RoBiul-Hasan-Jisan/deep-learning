#  Mini-Batch Gradient Descent 

## 1. What is Mini-Batch Gradient Descent?
**Mini-Batch Gradient Descent (MBGD)** is the ultimate hybrid of Batch Gradient Descent and Stochastic Gradient Descent (SGD). 

* Instead of using all the training data at once (Batch GD) or just one single sample at a time (SGD)...
* **MBGD uses a small, fixed-size subset** (a "mini-batch") of the data to compute the gradients and update the weights.

> **Key Idea:** Think of it as updating the network's weights after reviewing a small, manageable group of samples, rather than waiting for the whole dataset or reacting to every single data point.

---

## 2. How It Works
The process breaks down the training data into bite-sized chunks:

1. **Split the Data:** Divide the entire training dataset into mini-batches (e.g., 32, 64, or 128 samples per batch).
2. **For each individual mini-batch:**
   * **a.** Compute the predictions (Forward Pass).
   * **b.** Compute the loss.
   * **c.** Compute the gradients with respect to the weights and biases (Backpropagation).
   * **d.** Update the weights and biases using the gradients and the learning rate.
3. **Complete 1 Epoch:** Repeat step 2 for all mini-batches until you have gone through the entire dataset once.
4. **Repeat:** Do this for multiple epochs until the loss converges to the minimum.

---

## 3. Advantages of Mini-Batch Gradient Descent

| Feature | Mini-Batch GD Performance | Notes |
| :--- | :--- | :--- |
| **Speed** | Faster than Batch GD | Weight updates happen much more frequently. |
| **Stability** | Smoother than SGD | Averaging the gradient over a batch reduces extremely noisy, erratic updates. |
| **Memory** | Highly Efficient | Requires much less memory than Batch GD because it processes smaller chunks at a time. |
| **Convergence** | The Best Trade-off | Achieves the "sweet spot" between speed and stability. It is the gold standard for deep learning. |

**Important Note on Batch Size:**
Batch size is a crucial hyperparameter. Common values are `32`, `64`, `128`, or `256`.
* **Too small** $\rightarrow$ Becomes highly noisy, acting just like pure SGD.
* **Too large** $\rightarrow$ Becomes incredibly slow and memory-heavy, acting like Batch GD.

## 5. Key Takeaways
 **The Best of Both Worlds:** Mini-Batch GD perfectly balances the speed of SGD with the stability of Batch GD.  
 **Hyperparameter Tuning:** The Batch Size is highly important; `32` to `128` are the most common starting points in practice.  
 **Variance Reduction:** It significantly reduces the variance in weight updates, leading to a much smoother convergence curve.  
 **Industry Standard:** This is the default training method used under the hood in deep learning frameworks like **TensorFlow** and **PyTorch**.  
 **Highly Compatible:** It is almost always combined with advanced techniques like Momentum, Adam, or RMSProp for peak performance!

```python
import numpy as np

# Data
X = np.array([1,2,3,4,5,6])
Y = np.array([2,4,6,8,10,12])

# Initialize weight
w = 0.0
lr = 0.01
epochs = 5
batch_size = 2  # mini-batch size

# Mean Squared Error Loss
def compute_loss(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

# Training with Mini-Batch Gradient Descent
for _ in range(epochs):
    # Shuffle data
    indices = np.arange(len(X))
    np.random.shuffle(indices)
    X, Y = X[indices], Y[indices]
    
    # Mini-batches
    for i in range(0, len(X), batch_size):
        X_batch = X[i:i+batch_size]
        Y_batch = Y[i:i+batch_size]
        
        # Forward pass
        y_pred = w * X_batch
        # Gradient
        gradient = -2 * np.mean(X_batch * (Y_batch - y_pred))
        # Update weight
        w = w - lr * gradient
        
print("Trained weight:", w)
```
